```python
import os
from collections import defaultdict

output_dir = r"C:\Users\carlo\Documents\Algoritmos\yolooo\charts"
directory = output_dir

# Mapeamento de IDs para classes
id_to_class = {
    0: "engolfo_alta",
    1: "engolfo_baixa",
    2: "martelo",
    3: "pullback",
    4: "golden cross",
    5: "death cross"
}

class_count = defaultdict(int)
total_files = 0
unclassified = 0
unclassified_files = []

print(f"Processando diretório: {directory}")

for filename in os.listdir(directory):
    if filename.startswith("chart_") and filename.endswith(".txt"):
        total_files += 1
        filepath = os.path.join(directory, filename)
        
        try:
            with open(filepath, 'r') as file:
                # Lê apenas o primeiro número do arquivo
                first_number = file.read().strip().split()[0]
                
                try:
                    class_id = int(float(first_number))  # Converte para inteiro
                    
                    if class_id in id_to_class:
                        class_name = id_to_class[class_id]
                        class_count[class_name] += 1
                    else:
                        unclassified += 1
                        unclassified_files.append(filename)
                        print(f"ID não mapeado: {class_id} no arquivo {filename}")
                        
                except (ValueError, IndexError):
                    unclassified += 1
                    unclassified_files.append(filename)
                    print(f"Valor inválido: '{first_number}' no arquivo {filename}")
                    
        except Exception as e:
            print(f"Erro em {filename}: {str(e)}")
            unclassified += 1
            unclassified_files.append(filename)

# Relatório final
print("\n" + "="*50)
print("RELATÓRIO DE CLASSIFICAÇÃO")
print("="*50)
print(f"Total de arquivos processados: {total_files}")
print(f"Arquivos classificados: {total_files - unclassified}")
print(f"Arquivos não classificados: {unclassified}")

if unclassified > 0:
    print("\nArquivos não classificados (amostra):")
    for file in unclassified_files[:10]:
        print(f"- {file}")
    if unclassified > 10:
        print(f"... e mais {unclassified - 10} arquivos")

print("\nDistribuição de classes:")
for class_id, class_name in id_to_class.items():
    count = class_count[class_name]
    percentage = (count / total_files * 100) if total_files > 0 else 0
    print(f"- {class_name} (ID: {class_id}): {count} ocorrências | {percentage:.2f}%")

print("="*50)
```
