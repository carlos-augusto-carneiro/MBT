# Notebook para automatizar a coleta de dados EUR/USD e executar o notebook mbt.ipynb ao final

Este notebook realiza a automação da coleta de dados históricos do par EUR/USD utilizando Selenium, e ao final executa automaticamente o notebook `mbt.ipynb`.

## 1. Verificar e criar config.ini

Verifique se o arquivo `config.ini` existe no diretório atual. Caso não exista, crie o arquivo com as chaves necessárias para as credenciais de acesso ao Myfxbook.

```python
import os

config_path = 'config.ini'

if not os.path.exists(config_path):
    with open(config_path, 'w') as f:
        f.write("[MYFXBOOK]\n")
        f.write("EMAIL = seu_email@exemplo.com\n")
        f.write("PASSWORD = sua_senha\n")
    print("Arquivo config.ini criado. Por favor, adicione suas credenciais antes de continuar.")
else:
    print("Arquivo config.ini já existe.")
```

## 2. Ler e validar credenciais

Leia o arquivo `config.ini` e valide se as credenciais foram preenchidas corretamente antes de prosseguir com a coleta de dados.

```python
import configparser

config = configparser.ConfigParser()
config.read(config_path)

try:
    email = config['MYFXBOOK']['EMAIL']
    password = config['MYFXBOOK']['PASSWORD']
    if email == 'seu_email@exemplo.com' or password == 'sua_senha':
        raise ValueError("Por favor, atualize o arquivo config.ini com suas credenciais reais.")
    print("Credenciais encontradas e validadas.")
except Exception as e:
    print(f"Erro ao validar credenciais: {e}")
    raise
```

## 3. Coletar dados EUR/USD

Execute a função `fetch_eurusd_data()` para coletar e processar os dados históricos do par EUR/USD. Certifique-se de que todas as dependências estejam instaladas.

```python
# Importações necessárias (ajuste o caminho se necessário)
import sys
sys.path.append('.')

from MBTInfinity import fetch_eurusd_data

print("Iniciando processo de coleta de dados EUR/USD...")
df = fetch_eurusd_data()
if df is not None:
    print("Coleta de dados concluída com sucesso!")
else:
    print("Falha na coleta de dados.")
```

## 4. Executar notebook mbt.ipynb após o término

Utilize o módulo `subprocess` ou `nbclient` para executar automaticamente o notebook `mbt.ipynb` após o término da coleta de dados.

```python
# Usando nbclient para executar o notebook mbt.ipynb
import nbformat
from nbclient import NotebookClient

notebook_path = "MBT.ipynb"
notebook_path_2 = "MBT Analises.ipynb"

try:
    print(f"Executando notebook {notebook_path} ...")
    with open(notebook_path, encoding="utf-8") as f:        
        nb = nbformat.read(f, as_version=4)
    client = NotebookClient(nb, timeout=600, kernel_name="python3")
    client.execute()
    print("Execução do notebook mbt.ipynb concluída com sucesso.")
except Exception as e:
    print(f"Erro ao executar o notebook: {e}")

# Executar o segundo notebook
try:
    print(f"Executando notebook {notebook_path_2} ...")
    with open(notebook_path_2, encoding="utf-8") as f:        
        nb2 = nbformat.read(f, as_version=4)
    client2 = NotebookClient(nb2, timeout=600, kernel_name="python3")
    client2.execute()
    print("Execução do notebook MBT Analises.ipynb concluída com sucesso.")
except Exception as e:
    print(f"Erro ao executar o segundo notebook: {e}")
```
